```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from scipy.spatial import cKDTree

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import albumentations as A
from albumentations.pytorch import ToTensorV2

# Set reproducible seeds
torch.manual_seed(42)
np.random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using compute device:", device)

class KvasirDataset(Dataset):
    def __init__(self, data_dir, split_file, transform=None):
        self.data_dir = data_dir
        self.transform = transform
        
        split_path = os.path.join(data_dir, split_file)
        with open(split_path, "r") as f:
            self.image_ids = [line.strip() for line in f if line.strip()]

    def __len__(self):
        return len(self.image_ids)

    def __getitem__(self, idx):
        img_id = self.image_ids[idx]
        img_path = os.path.join(self.data_dir, "images", f"{img_id}.jpg")
        mask_path = os.path.join(self.data_dir, "masks", f"{img_id}.jpg")

        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        mask = (mask > 128).astype(np.float32)

        if self.transform is not None:
            augmented = self.transform(image=image, mask=mask)
            image = augmented["image"]
            mask = augmented["mask"]
        else:
            image = ToTensorV2()(image=image)["image"].float() / 255.0
            mask = torch.tensor(mask, dtype=torch.float32)

        if mask.ndim == 2:
            mask = mask.unsqueeze(0)
            
        return image, mask

def get_dataloaders(data_dir="data", batch_size=4, img_size=256):
    train_transform = A.Compose([
        A.Resize(img_size, img_size),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.RandomRotate90(p=0.5),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])
    
    val_transform = A.Compose([
        A.Resize(img_size, img_size),
        A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ToTensorV2(),
    ])
    
    train_dataset = KvasirDataset(data_dir, "train.txt", transform=train_transform)
    val_dataset = KvasirDataset(data_dir, "val.txt", transform=val_transform)
    
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=0)
    
    return train_loader, val_loader

train_loader, val_loader = get_dataloaders(data_dir="data", batch_size=4)
print(f"Data ready: {len(train_loader.dataset)} training samples, {len(val_loader.dataset)} validation samples.")
```

```python
class ConvBlock(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.conv(x)

# 1. Baseline U-Net
class UNet(nn.Module):
    def __init__(self, in_channels=3, num_classes=1):
        super().__init__()
        self.enc1 = ConvBlock(in_channels, 64)
        self.enc2 = ConvBlock(64, 128)
        self.enc3 = ConvBlock(128, 256)
        self.enc4 = ConvBlock(256, 512)
        
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.bottleneck = ConvBlock(512, 1024)
        
        self.up4 = nn.ConvTranspose2d(1024, 512, kernel_size=2, stride=2)
        self.dec4 = ConvBlock(1024, 512)
        
        self.up3 = nn.ConvTranspose2d(512, 256, kernel_size=2, stride=2)
        self.dec3 = ConvBlock(512, 256)
        
        self.up2 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.dec2 = ConvBlock(256, 128)
        
        self.up1 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.dec1 = ConvBlock(128, 64)
        
        self.final_conv = nn.Conv2d(64, num_classes, kernel_size=1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        
        b = self.bottleneck(self.pool(e4))
        
        d4 = self.up4(b)
        d4 = self.dec4(torch.cat([d4, e4], dim=1))
        
        d3 = self.up3(d4)
        d3 = self.dec3(torch.cat([d3, e3], dim=1))
        
        d2 = self.up2(d3)
        d2 = self.dec2(torch.cat([d2, e2], dim=1))
        
        d1 = self.up1(d2)
        d1 = self.dec1(torch.cat([d1, e1], dim=1))
        
        return self.final_conv(d1)

# 2. Attention U-Net
class AttentionGate(nn.Module):
    def __init__(self, f_g, f_l, f_int):
        super().__init__()
        self.w_g = nn.Sequential(
            nn.Conv2d(f_g, f_int, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(f_int)
        )
        self.w_x = nn.Sequential(
            nn.Conv2d(f_l, f_int, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(f_int)
        )
        self.psi = nn.Sequential(
            nn.Conv2d(f_int, 1, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(1),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU(inplace=True)

    def forward(self, g, x):
        g1 = self.w_g(g)
        x1 = self.w_x(x)
        psi = self.relu(g1 + x1)
        psi = self.psi(psi)
        return x * psi

class AttentionUNet(nn.Module):
    def __init__(self, in_channels=3, num_classes=1):
        super().__init__()
        self.enc1 = ConvBlock(in_channels, 64)
        self.enc2 = ConvBlock(64, 128)
        self.enc3 = ConvBlock(128, 256)
        self.enc4 = ConvBlock(256, 512)
        
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.bottleneck = ConvBlock(512, 1024)
        
        self.up4 = nn.ConvTranspose2d(1024, 512, kernel_size=2, stride=2)
        self.att4 = AttentionGate(f_g=512, f_l=512, f_int=256)
        self.dec4 = ConvBlock(1024, 512)
        
        self.up3 = nn.ConvTranspose2d(512, 256, kernel_size=2, stride=2)
        self.att3 = AttentionGate(f_g=256, f_l=256, f_int=128)
        self.dec3 = ConvBlock(512, 256)
        
        self.up2 = nn.ConvTranspose2d(256, 128, kernel_size=2, stride=2)
        self.att2 = AttentionGate(f_g=128, f_l=128, f_int=64)
        self.dec2 = ConvBlock(256, 128)
        
        self.up1 = nn.ConvTranspose2d(128, 64, kernel_size=2, stride=2)
        self.att1 = AttentionGate(f_g=64, f_l=64, f_int=32)
        self.dec1 = ConvBlock(128, 64)
        
        self.final_conv = nn.Conv2d(64, num_classes, kernel_size=1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        
        b = self.bottleneck(self.pool(e4))
        
        d4 = self.up4(b)
        x4 = self.att4(g=d4, x=e4)
        d4 = self.dec4(torch.cat([d4, x4], dim=1))
        
        d3 = self.up3(d4)
        x3 = self.att3(g=d3, x=e3)
        d3 = self.dec3(torch.cat([d3, x3], dim=1))
        
        d2 = self.up2(d3)
        x2 = self.att2(g=d2, x=e2)
        d2 = self.dec2(torch.cat([d2, x2], dim=1))
        
        d1 = self.up1(d2)
        x1 = self.att1(g=d1, x=e1)
        d1 = self.dec1(torch.cat([d1, x1], dim=1))
        
        return self.final_conv(d1)

print("UNet and AttentionUNet architectures defined successfully.")
```

```python
class DiceBCELoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth
        self.bce = nn.BCEWithLogitsLoss()

    def forward(self, preds, targets):
        bce_loss = self.bce(preds, targets)
        probs = torch.sigmoid(preds).view(-1)
        targets_flat = targets.view(-1)
        
        intersection = (probs * targets_flat).sum()
        dice_loss = 1.0 - (2.0 * intersection + self.smooth) / (probs.sum() + targets_flat.sum() + self.smooth)
        return bce_loss + dice_loss

def calculate_precision_recall(preds, masks):
    preds_flat = preds.view(-1)
    masks_flat = masks.view(-1)
    
    tp = (preds_flat * masks_flat).sum().item()
    fp = (preds_flat * (1.0 - masks_flat)).sum().item()
    fn = ((1.0 - preds_flat) * masks_flat).sum().item()
    
    precision = tp / (tp + fp + 1e-7)
    recall = tp / (tp + fn + 1e-7)
    return precision, recall

def calculate_hd95(preds, masks):
    preds_np = preds.detach().cpu().numpy().squeeze()
    masks_np = masks.detach().cpu().numpy().squeeze()
    
    if np.sum(preds_np) == 0 or np.sum(masks_np) == 0:
        return np.nan
        
    pred_coords = np.array(np.nonzero(preds_np)).T
    mask_coords = np.array(np.nonzero(masks_np)).T
    
    pred_tree = cKDTree(pred_coords)
    mask_tree = cKDTree(mask_coords)
    
    dist_pred_to_mask, _ = mask_tree.query(pred_coords)
    dist_mask_to_pred, _ = pred_tree.query(mask_coords)
    
    all_distances = np.concatenate([dist_pred_to_mask, dist_mask_to_pred])
    return np.percentile(all_distances, 95)

def evaluate_full_metrics(model, loader, device):
    model.eval()
    total_dice = 0.0
    total_iou = 0.0
    total_prec = 0.0
    total_rec = 0.0
    hd95_list = []
    
    with torch.no_grad():
        for images, masks in loader:
            images = images.to(device)
            masks = masks.to(device)
            
            outputs = model(images)
            preds = (torch.sigmoid(outputs) > 0.5).float()
            
            intersection = (preds * masks).sum()
            union = preds.sum() + masks.sum()
            
            total_dice += (2.0 * intersection / (union + 1e-7)).item()
            total_iou += (intersection / (union - intersection + 1e-7)).item()
            
            prec, rec = calculate_precision_recall(preds, masks)
            total_prec += prec
            total_rec += rec
            
            for p, m in zip(preds, masks):
                val_hd95 = calculate_hd95(p, m)
                if not np.isnan(val_hd95):
                    hd95_list.append(val_hd95)
                    
    num_batches = len(loader)
    return {
        "Dice": total_dice / num_batches,
        "IoU": total_iou / num_batches,
        "Precision": total_prec / num_batches,
        "Recall": total_rec / num_batches,
        "HD95": float(np.mean(hd95_list)) if len(hd95_list) > 0 else np.nan
    }

print("Evaluation engine loaded.")
```

```python
print("running training") 
def train_model(model, train_loader, val_loader, criterion, optimizer, epochs=15, save_name="model.pth"):
    best_dice = 0.0
    for epoch in range(1, epochs + 1):
        model.train()
        train_loss = 0.0
        for images, masks in tqdm(train_loader, desc=f"Epoch {epoch}/{epochs} [Train]", leave=False):
            images, masks = images.to(device), masks.to(device)
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, masks)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * images.size(0)
        
        train_loss /= len(train_loader.dataset)
        
        # Validation
        model.eval()
        val_loss = 0.0
        val_dice = 0.0
        with torch.no_grad():
            for images, masks in val_loader:
                images, masks = images.to(device), masks.to(device)
                outputs = model(images)
                loss = criterion(outputs, masks)
                val_loss += loss.item() * images.size(0)
                
                preds = (torch.sigmoid(outputs) > 0.5).float()
                intersection = (preds * masks).sum().item()
                val_dice += (2.0 * intersection / (preds.sum().item() + masks.sum().item() + 1e-7))
                
        val_loss /= len(val_loader.dataset)
        val_dice /= len(val_loader)
        
        print(f"Epoch {epoch:02d}/{epochs:02d} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Val Dice: {val_dice:.4f}")
        
        if val_dice > best_dice:
            best_dice = val_dice
            torch.save(model.state_dict(), save_name)
            print(f"  -> Checkpoint saved: {save_name} (Dice: {best_dice:.4f})")

criterion = DiceBCELoss()

# Train Baseline U-Net if best_unet.pth is not yet created
if not os.path.exists("best_unet.pth"):
    print("\n--- Training Baseline U-Net ---")
    unet_model = UNet(in_channels=3, num_classes=1).to(device)
    unet_opt = torch.optim.Adam(unet_model.parameters(), lr=1e-4)
    train_model(unet_model, train_loader, val_loader, criterion, unet_opt, epochs=15, save_name="best_unet.pth")
else:
    print("Found existing 'best_unet.pth'.")

# Train Attention U-Net if best_att_unet.pth is not present
if not os.path.exists("best_att_unet.pth"):
    print("\n--- Training Attention U-Net ---")
    att_model = AttentionUNet(in_channels=3, num_classes=1).to(device)
    att_opt = torch.optim.Adam(att_model.parameters(), lr=1e-4)
    train_model(att_model, train_loader, val_loader, criterion, att_opt, epochs=15, save_name="best_att_unet.pth")
else:
    print("Found existing 'best_att_unet.pth'.")
```

```python
# Instantiate clean architectures
baseline_unet = UNet(in_channels=3, num_classes=1).to(device)
attention_unet = AttentionUNet(in_channels=3, num_classes=1).to(device)

# Load checkpoint weights
baseline_unet.load_state_dict(torch.load("best_unet.pth", map_location=device, weights_only=True))
attention_unet.load_state_dict(torch.load("best_att_unet.pth", map_location=device, weights_only=True))

print("Evaluating Baseline U-Net on clean data...")
unet_clean_metrics = evaluate_full_metrics(baseline_unet, val_loader, device)

print("Evaluating Attention U-Net on clean data...")
att_clean_metrics = evaluate_full_metrics(attention_unet, val_loader, device)

# Print comparative results table for your IEEE report
print("\n" + "="*58)
print(" TASK 3: OFFICIAL PERFORMANCE COMPARISON (CLEAN DATA)")
print("="*58)
print(f"{'Metric':<14} | {'Baseline U-Net':<18} | {'Attention U-Net':<18}")
print("-" * 58)
for m in ["Dice", "IoU", "Precision", "Recall", "HD95"]:
    print(f"{m:<14} | {unet_clean_metrics[m]:<18.4f} | {att_clean_metrics[m]:<18.4f}")
print("="*58)
```

```python
def apply_corruption(images, corruption_type):
    corrupted = images.clone()
    
    if corruption_type == "gaussian_noise_5":
        # 5% additive Gaussian noise relative to standard variance (0.05 std)
        noise = torch.randn_like(corrupted) * 0.05
        corrupted = corrupted + noise
        
    elif corruption_type == "gaussian_noise_10":
        # 10% additive Gaussian noise relative to standard variance (0.10 std)
        noise = torch.randn_like(corrupted) * 0.10
        corrupted = corrupted + noise
        
    elif corruption_type == "blur":
        # Approximate 5x5 Gaussian blur via low-pass depthwise convolution
        kernel_1d = torch.tensor([1, 4, 6, 4, 1], dtype=torch.float32, device=images.device) / 16.0
        kernel_2d = (kernel_1d[:, None] * kernel_1d[None, :]).unsqueeze(0).unsqueeze(0)
        kernel_2d = kernel_2d.repeat(3, 1, 1, 1)
        corrupted = nn.functional.conv2d(corrupted, kernel_2d, padding=2, groups=3)
        
    elif corruption_type == "contrast_change":
        # Alter image contrast by scaling deviations around mean
        mean = corrupted.mean(dim=(2, 3), keepdim=True)
        corrupted = (corrupted - mean) * 0.6 + mean
        
    elif corruption_type == "intensity_norm_shift":
        # Simulates distribution shift / uncalibrated acquisition differences
        shift = 0.5
        corrupted = (corrupted + shift) * 0.8

    return corrupted

def evaluate_robustness(model, loader, device, corruption_type):
    model.eval()
    total_dice = 0.0
    total_iou = 0.0
    
    with torch.no_grad():
        for images, masks in loader:
            images = images.to(device)
            masks = masks.to(device)
            
            # Corrupt batch
            corrupted_images = apply_corruption(images, corruption_type)
            outputs = model(corrupted_images)
            preds = (torch.sigmoid(outputs) > 0.5).float()
            
            intersection = (preds * masks).sum()
            union = preds.sum() + masks.sum()
            
            total_dice += (2.0 * intersection / (union + 1e-7)).item()
            total_iou += (intersection / (union - intersection + 1e-7)).item()
            
    num_batches = len(loader)
    return total_dice / num_batches, total_iou / num_batches

# Evaluate both models across all corruptions
corruptions = [
    "gaussian_noise_5",
    "gaussian_noise_10",
    "blur",
    "contrast_change",
    "intensity_norm_shift"
]

print("Running Task 4 & 5: Robustness Evaluations...")
robustness_summary = []

for c in corruptions:
    u_dice, u_iou = evaluate_robustness(baseline_unet, val_loader, device, c)
    a_dice, a_iou = evaluate_robustness(attention_unet, val_loader, device, c)
    robustness_summary.append((c, u_dice, a_dice, u_iou, a_iou))

print("\n" + "="*80)
print(" TASK 5: ROBUSTNESS COMPARISON BEFORE AND AFTER CORRUPTIONS")
print("="*80)
print(f"{'Condition':<24} | {'Baseline Dice':<14} | {'Att-UNet Dice':<14} | {'Baseline IoU':<14} | {'Att-UNet IoU':<14}")
print("-" * 80)
print(f"{'Clean (Baseline)':<24} | {unet_clean_metrics['Dice']:<14.4f} | {att_clean_metrics['Dice']:<14.4f} | {unet_clean_metrics['IoU']:<14.4f} | {att_clean_metrics['IoU']:<14.4f}")
for name, u_d, a_d, u_i, a_i in robustness_summary:
    print(f"{name:<24} | {u_d:<14.4f} | {a_d:<14.4f} | {u_i:<14.4f} | {a_i:<14.4f}")
print("="*80)
```

```python
def visualize_comparative_qualitative(baseline_model, att_model, loader, device, num_samples=2):
    baseline_model.eval()
    att_model.eval()
    
    images, masks = next(iter(loader))
    images = images[:num_samples].to(device)
    masks = masks[:num_samples].to(device)
    
    # Generate corrupted versions (Gaussian noise 10% and blur)
    noise_imgs = apply_corruption(images, "gaussian_noise_10")
    blur_imgs = apply_corruption(images, "blur")
    
    with torch.no_grad():
        clean_pred_base = (torch.sigmoid(baseline_model(images)) > 0.5).float()
        clean_pred_att = (torch.sigmoid(att_model(images)) > 0.5).float()
        
        noise_pred_base = (torch.sigmoid(baseline_model(noise_imgs)) > 0.5).float()
        noise_pred_att = (torch.sigmoid(att_model(noise_imgs)) > 0.5).float()
        
        blur_pred_att = (torch.sigmoid(att_model(blur_imgs)) > 0.5).float()

    images_np = images.cpu().numpy()
    masks_np = masks.cpu().numpy()
    clean_pred_base_np = clean_pred_base.cpu().numpy()
    clean_pred_att_np = clean_pred_att.cpu().numpy()
    noise_pred_base_np = noise_pred_base.cpu().numpy()
    noise_pred_att_np = noise_pred_att.cpu().numpy()

    for idx in range(num_samples):
        fig, axs = plt.subplots(1, 6, figsize=(20, 4))
        
        # Display image
        img = np.transpose(images_np[idx], (1, 2, 0))
        img = (img - img.min()) / (img.max() - img.min() + 1e-7)
        
        axs[0].imshow(img)
        axs[0].set_title("Input Image")
        axs[0].axis("off")
        
        axs[1].imshow(masks_np[idx].squeeze(), cmap="gray")
        axs[1].set_title("Ground Truth")
        axs[1].axis("off")
        
        axs[2].imshow(clean_pred_base_np[idx].squeeze(), cmap="gray")
        axs[2].set_title("U-Net (Clean)")
        axs[2].axis("off")
        
        axs[3].imshow(clean_pred_att_np[idx].squeeze(), cmap="gray")
        axs[3].set_title("Att-UNet (Clean)")
        axs[3].axis("off")
        
        axs[4].imshow(noise_pred_base_np[idx].squeeze(), cmap="gray")
        axs[4].set_title("U-Net (10% Noise)")
        axs[4].axis("off")
        
        axs[5].imshow(noise_pred_att_np[idx].squeeze(), cmap="gray")
        axs[5].set_title("Att-UNet (10% Noise)")
        axs[5].axis("off")
        
        plt.tight_layout()
        plt.show()

visualize_comparative_qualitative(baseline_unet, attention_unet, val_loader, device)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torch

def visualize_any_samples(baseline_model, att_model, dataset, device, sample_indices=[0, 5, 10, 15, 20]):
  
    baseline_model.eval()
    att_model.eval()
    
    for sample_idx in sample_indices:
        # Fetch individual sample directly from dataset
        image_tensor, mask_tensor = dataset[sample_idx]
        
        # Add batch dimension: [1, C, H, W]
        img_batch = image_tensor.unsqueeze(0).to(device)
        mask_batch = mask_tensor.unsqueeze(0).to(device)
        
        # Apply synthetic corruptions to this specific image
        blur_batch = apply_corruption(img_batch, "blur")
        noise_batch = apply_corruption(img_batch, "gaussian_noise_10")
        
        with torch.no_grad():
            clean_pred_base = (torch.sigmoid(baseline_model(img_batch)) > 0.5).float()
            clean_pred_att = (torch.sigmoid(att_model(img_batch)) > 0.5).float()
            
            blur_pred_base = (torch.sigmoid(baseline_model(blur_batch)) > 0.5).float()
            blur_pred_att = (torch.sigmoid(att_model(blur_batch)) > 0.5).float()

        # Format arrays for plotting
        img_np = img_batch.cpu().squeeze().numpy().transpose(1, 2, 0)
        img_np = (img_np - img_np.min()) / (img_np.max() - img_np.min() + 1e-7)
        
        mask_np = mask_batch.cpu().squeeze().numpy()
        pred_base_np = clean_pred_base.cpu().squeeze().numpy()
        pred_att_np = clean_pred_att.cpu().squeeze().numpy()
        blur_base_np = blur_pred_base.cpu().squeeze().numpy()
        blur_att_np = blur_pred_att.cpu().squeeze().numpy()

        fig, axs = plt.subplots(1, 6, figsize=(18, 3))
        fig.suptitle(f"Validation Sample #{sample_idx} (Image ID: {dataset.image_ids[sample_idx]})", fontsize=12, y=1.05)
        
        axs[0].imshow(img_np)
        axs[0].set_title("Input Image")
        axs[0].axis("off")
        
        axs[1].imshow(mask_np, cmap="gray")
        axs[1].set_title("Ground Truth")
        axs[1].axis("off")
        
        axs[2].imshow(pred_base_np, cmap="gray")
        axs[2].set_title("U-Net (Clean)")
        axs[2].axis("off")
        
        axs[3].imshow(pred_att_np, cmap="gray")
        axs[3].set_title("Att-UNet (Clean)")
        axs[3].axis("off")
        
        axs[4].imshow(blur_base_np, cmap="gray")
        axs[4].set_title("U-Net (Blur)")
        axs[4].axis("off")
        
        axs[5].imshow(blur_att_np, cmap="gray")
        axs[5].set_title("Att-UNet (Blur)")
        axs[5].axis("off")
        
        plt.tight_layout()
        plt.show()

# --- RUN: Pick any indices you want from val_dataset ---
# For instance, inspect 5 different images spread across the validation set:
chosen_indices = [2, 8, 14, 25, 30]

visualize_any_samples(baseline_unet, attention_unet, val_loader.dataset, device, sample_indices=chosen_indices)
```

```python
import os
import cv2
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
import albumentations as A
from albumentations.pytorch import ToTensorV2
from torch.utils.data import Dataset, DataLoader

# -------------------------------------------------------------
# 1. ARCHITECTURE DEFINITIONS
# -------------------------------------------------------------
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
            nn.ReLU(inplace=True)
        )
    def forward(self, x):
        return self.conv(x)

class AttentionGate(nn.Module):
    def __init__(self, F_g, F_l, F_int):
        super().__init__()
        self.W_g = nn.Sequential(
            nn.Conv2d(F_g, F_int, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(F_int)
        )
        self.W_x = nn.Sequential(
            nn.Conv2d(F_l, F_int, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(F_int)
        )
        self.psi = nn.Sequential(
            nn.Conv2d(F_int, 1, kernel_size=1, stride=1, padding=0, bias=True),
            nn.BatchNorm2d(1),
            nn.Sigmoid()
        )
        self.relu = nn.ReLU(inplace=True)

    def forward(self, g, x):
        g1 = self.W_g(g)
        x1 = self.W_x(x)
        psi = self.relu(g1 + x1)
        psi = self.psi(psi)
        return x * psi

class UNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1, features=[64, 128, 256, 512]):
        super().__init__()
        self.downs = nn.ModuleList()
        self.ups = nn.ModuleList()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        curr_in = in_channels
        for feature in features:
            self.downs.append(DoubleConv(curr_in, feature))
            curr_in = feature

        self.bottleneck = DoubleConv(features[-1], features[-1] * 2)

        for feature in reversed(features):
            self.ups.append(nn.ConvTranspose2d(feature * 2, feature, kernel_size=2, stride=2))
            self.ups.append(DoubleConv(feature * 2, feature))

        self.final_conv = nn.Conv2d(features[0], out_channels, kernel_size=1)

    def forward(self, x):
        skip_connections = []
        for down in self.downs:
            x = down(x)
            skip_connections.append(x)
            x = self.pool(x)

        x = self.bottleneck(x)
        skip_connections = skip_connections[::-1]

        for idx in range(0, len(self.ups), 2):
            x = self.ups[idx](x)
            skip_conn = skip_connections[idx // 2]
            if x.shape != skip_conn.shape:
                x = F.interpolate(x, size=skip_conn.shape[2:], mode="bilinear", align_corners=True)
            x = torch.cat((skip_conn, x), dim=1)
            x = self.ups[idx + 1](x)

        return self.final_conv(x)

class AttentionUNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1, features=[64, 128, 256, 512]):
        super().__init__()
        self.downs = nn.ModuleList()
        self.ups = nn.ModuleList()
        self.gates = nn.ModuleList()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        curr_in = in_channels
        for feature in features:
            self.downs.append(DoubleConv(curr_in, feature))
            curr_in = feature

        self.bottleneck = DoubleConv(features[-1], features[-1] * 2)

        for feature in reversed(features):
            self.ups.append(nn.ConvTranspose2d(feature * 2, feature, kernel_size=2, stride=2))
            self.gates.append(AttentionGate(F_g=feature, F_l=feature, F_int=feature // 2))
            self.ups.append(DoubleConv(feature * 2, feature))

        self.final_conv = nn.Conv2d(features[0], out_channels, kernel_size=1)

    def forward(self, x):
        skip_connections = []
        for down in self.downs:
            x = down(x)
            skip_connections.append(x)
            x = self.pool(x)

        x = self.bottleneck(x)
        skip_connections = skip_connections[::-1]

        for i in range(len(self.gates)):
            x = self.ups[i * 2](x)
            skip_conn = skip_connections[i]
            if x.shape != skip_conn.shape:
                x = F.interpolate(x, size=skip_conn.shape[2:], mode="bilinear", align_corners=True)
            gated_skip = self.gates[i](g=x, x=skip_conn)
            x = torch.cat((gated_skip, x), dim=1)
            x = self.ups[i * 2 + 1](x)

        return self.final_conv(x)

# -------------------------------------------------------------
# 2. DATASET & CORRUPTION HELPER
# -------------------------------------------------------------
class PolypValDataset(Dataset):
    def __init__(self, val_file, data_root="data", img_size=256):
        with open(val_file, "r") as f:
            self.image_ids = [line.strip() for line in f if line.strip()]
        self.data_root = data_root
        self.transform = A.Compose([
            A.Resize(img_size, img_size),
            A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
            ToTensorV2(),
        ])

    def __len__(self):
        return len(self.image_ids)

    def __getitem__(self, idx):
        file_name = self.image_ids[idx]
        img_path = os.path.join(self.data_root, "images", f"{file_name}.jpg")
        mask_path = os.path.join(self.data_root, "masks", f"{file_name}.jpg")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        mask = (mask > 128).astype(np.float32)

        augmented = self.transform(image=img, mask=mask)
        return augmented["image"], augmented["mask"].unsqueeze(0)

def apply_corruption(tensor_batch, ctype):
    t = tensor_batch.clone()
    if ctype == "blur":
        # 5x5 average blur kernel
        kernel = torch.ones((3, 1, 5, 5), device=tensor_batch.device) / 25.0
        return F.conv2d(t, kernel, padding=2, groups=3)
    return t

def compute_single_metrics(pred, target, eps=1e-7):
    pred = pred.astype(bool)
    target = target.astype(bool)
    intersection = np.logical_and(pred, target).sum()
    total = pred.sum() + target.sum()
    union = np.logical_or(pred, target).sum()
    dice = (2.0 * intersection + eps) / (total + eps)
    iou = (intersection + eps) / (union + eps)
    return dice, iou

# -------------------------------------------------------------
# 3. INITIALIZE, LOAD WEIGHTS, & PLOT
# -------------------------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

baseline_unet = UNet(in_channels=3, out_channels=1).to(device)
attention_unet = AttentionUNet(in_channels=3, out_channels=1).to(device)

if os.path.exists("best_unet.pth"):
    baseline_unet.load_state_dict(torch.load("best_unet.pth", map_location=device))
    print("Loaded best_unet.pth")

if os.path.exists("best_att_unet.pth"):
    attention_unet.load_state_dict(torch.load("best_att_unet.pth", map_location=device))
    print("Loaded best_att_unet.pth")

# Load validation split
val_dataset = PolypValDataset(val_file="data/val.txt", data_root="data")

# Generate quantitative sample image
baseline_unet.eval()
attention_unet.eval()

sample_idx = 8
image_tensor, mask_tensor = val_dataset[sample_idx]
img_batch = image_tensor.unsqueeze(0).to(device)
mask_np = mask_tensor.squeeze().cpu().numpy().astype(np.float32)

blur_batch = apply_corruption(img_batch, "blur")

with torch.no_grad():
    clean_base = (torch.sigmoid(baseline_unet(img_batch)) > 0.5).float().cpu().squeeze().numpy()
    clean_att  = (torch.sigmoid(attention_unet(img_batch)) > 0.5).float().cpu().squeeze().numpy()
    blur_base  = (torch.sigmoid(baseline_unet(blur_batch)) > 0.5).float().cpu().squeeze().numpy()
    blur_att   = (torch.sigmoid(attention_unet(blur_batch)) > 0.5).float().cpu().squeeze().numpy()

# Calculate numbers
d_c_base, i_c_base = compute_single_metrics(clean_base, mask_np)
d_c_att,  i_c_att  = compute_single_metrics(clean_att,  mask_np)
d_b_base, i_b_base = compute_single_metrics(blur_base,  mask_np)
d_b_att,  i_b_att  = compute_single_metrics(blur_att,   mask_np)

# Normalize for rendering
img_np = img_batch.cpu().squeeze().numpy().transpose(1, 2, 0)
img_np = (img_np - img_np.min()) / (img_np.max() - img_np.min() + 1e-7)

fig, axs = plt.subplots(1, 6, figsize=(20, 3.8))
fig.suptitle(f"Quantitative Sample #{sample_idx} ({val_dataset.image_ids[sample_idx]})", fontsize=13, y=1.03)

axs[0].imshow(img_np)
axs[0].set_title("Input Image\n(Clean)")
axs[0].axis("off")

axs[1].imshow(mask_np, cmap="gray")
axs[1].set_title("Ground Truth\n(Reference)")
axs[1].axis("off")

axs[2].imshow(clean_base, cmap="gray")
axs[2].set_title(f"U-Net (Clean)\nDice: {d_c_base:.3f} | IoU: {i_c_base:.3f}")
axs[2].axis("off")

axs[3].imshow(clean_att, cmap="gray")
axs[3].set_title(f"Att-UNet (Clean)\nDice: {d_c_att:.3f} | IoU: {i_c_att:.3f}")
axs[3].axis("off")

axs[4].imshow(blur_base, cmap="gray")
axs[4].set_title(f"U-Net (Blur)\nDice: {d_b_base:.3f} | IoU: {i_b_base:.3f}")
axs[4].axis("off")

axs[5].imshow(blur_att, cmap="gray")
axs[5].set_title(f"Att-UNet (Blur)\nDice: {d_b_att:.3f} | IoU: {i_b_att:.3f}")
axs[5].axis("off")

plt.tight_layout()
plt.savefig("quantitative_sample_8.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved quantitative_sample_8.png successfully!")
```
